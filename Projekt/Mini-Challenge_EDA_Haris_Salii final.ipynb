Danach baue alle Analysen auf diesem bereinigten Datensatz auf. Sortierte Kategorien zeige ich in allen Grafiken in der Reihenfolge Low, Medium, High, damit die Plots gut vergleichbar sind.</mark>

### Datenbereinigung und -vorbereitung

Bevor ich mit den Analysen starte, schaue ich mir die Daten an und bereinige sie:

1. **Dubletten prüfen:** <mark>Gibt es doppelte Zeilen?
2. **Fehlende Werte entfernen:** <mark>Nur drei Spalten (`Teacher_Quality`, `Parental_Education_Level`, `Distance_from_Home`) haben Lücken, und das nur bei wenigen der 6607 Zeilen. Ich lösche diese Zeilen, weil der Datenverlust klein ist und der Datensatz dadurch sauber bleibt.</mark>
3. **Ausreisser entfernen:** <mark>Der Exam_Score 101 ist unmöglich (das Maximum sind 100 Punkte), darum entferne ich diese eine Zeile.</mark>
4. **Kategorien ordnen:** <mark>Die sortierten Variablen (Low/Medium/High) setze ich als geordnete Kategorie, damit sie überall in der richtigen Reihenfolge erscheinen.</mark>
5. **Neue Spalte:** <mark>Ich teile die Lernstunden in die Gruppen wenig/mittel/viel ein. Diese Spalte verwende ich später in der Heatmap (Analyse 3).</mark>

```python
# 1. Dubletten prüfen
print("Anzahl Dubletten:", df.duplicated().sum())

# 2. Zeilen mit fehlenden Werten löschen
print("\nFehlende Werte pro Spalte (vorher):")
print(df.isna().sum()[df.isna().sum() > 0])
df = df.dropna()
print("Zeilen nach dem Löschen:", len(df))

# 3. Unmöglichen Ausreisser entfernen (Exam_Score über 100)
print("\nZeilen mit Exam_Score über 100:", (df["Exam_Score"] > 100).sum())
df = df[df["Exam_Score"] <= 100].copy()
print("Zeilen nach dem Entfernen:", len(df))

# 4. Sortierte Kategorien in die richtige Reihenfolge bringen
kat_reihenfolge = ["Low", "Medium", "High"]
for spalte in ["Parental_Involvement", "Access_to_Resources", "Motivation_Level",
               "Family_Income", "Teacher_Quality"]:
    df[spalte] = pd.Categorical(df[spalte], categories=kat_reihenfolge, ordered=True)

# 5. Neue Hilfsspalte: Lernstunden in Gruppen (wird in Analyse 3 gebraucht)
df["Hours_Kategorie"] = pd.cut(df["Hours_Studied"], bins=[0, 10, 20, 50],
                                labels=["wenig", "mittel", "viel"])

print("\nDaten sind bereit. Verbleibende Zeilen:", len(df))
```

**Analyse 1:**

1. <mark>Ich schaue, wie stark jede Variable mit dem Exam_Score zusammenhängt, also die numerischen und die in Zahlen umgewandelten kategorialen Variablen.</mark>
2. <mark>So bekomme ich einen Überblick, welche Faktoren ich genauer anschauen sollte. Genau das fehlt im automatischen Report bei den kategorialen Variablen.</mark>
3. <mark>Ein liegendes Balkendiagramm passt gut, weil ich viele benannte Faktoren sortiert untereinander zeigen kann. Positive und negative Werte sind links und rechts der Null leicht zu unterscheiden.</mark>
4. <mark>Ich wandle die sortierten Kategorien sinnvoll in Zahlen um (Low=1, Medium=2, High=3) und sortiere nach der Stärke. Bei Variablen ohne echte Reihenfolge wie Gender zählt nur, dass der Wert nahe Null liegt, das Vorzeichen sagt dort nichts aus.</mark>
5. <mark>Attendance (0.58) und Hours_Studied (0.45) sind klar am stärksten. Danach kommen mit Abstand Previous_Scores, Access_to_Resources, Parental_Involvement und Tutoring_Sessions (rund 0.16 bis 0.18). Gender, School_Type und Sleep_Hours haben fast keinen Zusammenhang. Das passt zu Hypothese 1.</mark>

```python
# Sortierte/kategoriale Werte in Zahlen umwandeln (nur für diese Korrelation)
codierung = {"Low": 1, "Medium": 2, "High": 3, "No": 0, "Yes": 1,
             "Public": 0, "Private": 1, "Negative": 1, "Neutral": 2, "Positive": 3,
             "Male": 0, "Female": 1, "Near": 1, "Moderate": 2, "Far": 3,
             "High School": 1, "College": 2, "Postgraduate": 3}

# Kopie ohne die Hilfsspalte; alle nicht-numerischen Spalten in Zahlen umwandeln
df_zahlen = df.drop(columns="Hours_Kategorie").copy()
for spalte in df_zahlen.select_dtypes(exclude="number").columns:
    df_zahlen[spalte] = df_zahlen[spalte].astype(str).map(codierung) # alle nicht-numerischen Spalten in Zahlen umwandeln

# Korrelation jeder Spalte mit dem Exam_Score, nach Stärke sortiert
korr = df_zahlen.corr(numeric_only=True)["Exam_Score"].drop("Exam_Score").sort_values()

korr.plot.barh(figsize=(9, 7), color="#3b6ea5") # horizontales Balkendiagramm
plt.title("Korrelation jedes Faktors mit dem Exam_Score")
plt.xlabel("Pearson-Korrelation")
plt.tight_layout()
plt.show()
```

**Analyse 2:**

1. <mark>Ich schaue mir den Zusammenhang zwischen der Anwesenheit (in Prozent) und dem Exam_Score genauer an.</mark>
2. <mark>Attendance war der stärkste Faktor. Ich will sehen, ob der Zusammenhang wirklich gleichmässig ansteigt und wie eng er ist.</mark>
3. <mark>Ein Streudiagramm zeigt zwei numerische Variablen direkt. Mit der zusätzlichen Geraden sieht man die Richtung des Zusammenhangs auf einen Blick.</mark>
4. <mark>Bei 6377 Punkten liegen viele übereinander. Ich mache die Punkte deshalb durchsichtig (alpha) und lasse seaborn automatisch eine Trendgerade einzeichnen.</mark>
5. <mark>Die Gerade steigt klar an: bei höherer Anwesenheit ist auch der Exam_Score höher. Die Punkte streuen aber stark, die Anwesenheit allein erklärt die Leistung also nicht vollständig.</mark>

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Attendance", y="Exam_Score",
            scatter_kws={"alpha": 0.1, "s": 10}, line_kws={"color": "red"}) # Streudiagramm mit Regressionslinie
plt.title("Anwesenheit vs. Exam_Score")
plt.xlabel("Anwesenheit (%)")
plt.ylabel("Exam_Score")
plt.tight_layout()
plt.show()
```

**Analyse 3:**

1. <mark>Ich schaue, wie Anwesenheit und Lernstunden zusammen wirken. Dafür teile ich die Anwesenheit in vier Bereiche ein und nutze für die Lernstunden die Gruppen wenig/mittel/viel aus der Datenvorbereitung. Pro Kombination berechne ich den durchschnittlichen Exam_Score.</mark>
2. <mark>Beide Faktoren sind einzeln stark (Analyse 1). Mich interessiert, ob sie sich zusammen verstärken. Das zeigt der automatische Report nicht.</mark>
3. <mark>Eine Heatmap ist gut geeignet, um einen Durchschnittswert über zwei Gruppen als Farbe darzustellen. So sieht man das Muster sofort.</mark>
4. <mark>Ich schreibe den Mittelwert in jedes Feld und nehme eine Farbskala, bei der dunkler einen höheren Score bedeutet.</mark>
5. <mark>Der Score steigt klar von links oben (wenig Anwesenheit, wenig Lernen) nach rechts unten (viel Anwesenheit, viel Lernen). Wer beides hat, erreicht die besten Ergebnisse. Die beiden Faktoren ergänzen sich also.</mark>

```python
# Anwesenheit in Bereiche einteilen; die Lernstunden-Gruppe kommt aus der Datenvorbereitung
anwesenheit = pd.cut(df["Attendance"], [59, 70, 80, 90, 100],
                     labels=["60-70", "71-80", "81-90", "91-100"])

# Durchschnittlichen Exam_Score je Kombination berechnen
tabelle = df.pivot_table(index=anwesenheit, columns="Hours_Kategorie",
                         values="Exam_Score", aggfunc="mean")

plt.figure(figsize=(7, 5))
sns.heatmap(tabelle, annot=True, fmt=".1f", cmap="Blues")
plt.title("Durchschnittlicher Exam_Score nach Anwesenheit und Lernstunden")
plt.xlabel("Lernstunden pro Woche")
plt.ylabel("Anwesenheit (%)")
plt.tight_layout()
plt.show()
```

**Analyse 4:**

1. <mark>Ich vergleiche den durchschnittlichen Exam_Score über die Stufen des Familieneinkommens, zusätzlich getrennt nach Ressourcenzugang.</mark>
2. <mark>Das gehört zu Fragestellung 2: Spielt der Hintergrund eine Rolle, und ist eher das Einkommen oder der Ressourcenzugang wichtig?</mark>
3. <mark>Ein gruppiertes Balkendiagramm eignet sich, um einen Durchschnitt über zwei Kategorien direkt nebeneinander zu vergleichen.</mark>
4. <mark>Ich sortiere die Stufen von Low bis High. Die y-Achse starte ich bei 64, sonst sieht man die kleinen Unterschiede nicht. Man muss aber im Kopf behalten, dass die Unterschiede in Wirklichkeit klein sind (etwa 1 bis 2 Punkte).</mark>
5. <mark>Innerhalb jeder Einkommensstufe steigt der Score mit besserem Ressourcenzugang (rund 1.5 Punkte von Low zu High). Das Einkommen selbst zeigt dagegen kaum einen klaren Trend. Der Effekt ist also da, aber klein. Das passt zu Hypothese 2.</mark>

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Family_Income", y="Exam_Score", hue="Access_to_Resources",
            order=kat_reihenfolge, hue_order=kat_reihenfolge, errorbar=None, palette="Blues") 
plt.ylim(64, 70)   # Achse gekürzt, damit man die kleinen Unterschiede sieht
plt.title("Durchschnittlicher Exam_Score nach Einkommen und Ressourcen")
plt.xlabel("Familieneinkommen")
plt.ylabel("Durchschnittlicher Exam_Score")
plt.legend(title="Ressourcenzugang")
plt.tight_layout()
plt.show()
```

**Analyse 5:**

1. <mark>Ich vergleiche die Verteilung des Exam_Score über die drei Stufen der elterlichen Einbindung (Parental_Involvement: Low, Medium, High).</mark>
2. <mark>In Analyse 4 sehe ich nur Mittelwerte. Hier will ich die ganze Verteilung sehen, also Median, Streuung und Ausreisser. Diese sagen oft mehr aus als der Mittelwert allein.</mark>
3. <mark>Ein Boxplot ist der Standard, um eine numerische Variable über wenige Gruppen zu vergleichen. Median, Streuung und Ausreisser sieht man gleichzeitig.</mark>
4. <mark>Ich sortiere die Gruppen von Low bis High. Die einzelnen Punkte ausserhalb der Boxen sind die Ausreisser.</mark>
5. <mark>Der Median steigt nur leicht: von 66 (Low) über 67 (Medium) zu 68 (High). Die Boxen überschneiden sich aber stark. Mehr Einbindung bedeutet also tendenziell etwas höhere, aber nicht sicher bessere Noten. Die vielen hohen Ausreisser zeigen, dass die Verteilung nach oben einen langen Rand hat.</mark>

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Parental_Involvement", y="Exam_Score",
            order=kat_reihenfolge, color="#3b6ea5")
plt.title("Verteilung des Exam_Score nach elterlicher Einbindung")
plt.xlabel("Elterliche Einbindung")
plt.ylabel("Exam_Score")
plt.tight_layout() # damit Titel und Achsenbeschriftungen nicht abgeschnitten werden
plt.show()
```
